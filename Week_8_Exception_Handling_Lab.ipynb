## Banking ATM System

### Requirements
- Create a `BankAccount` class.
- Add `deposit()`, `withdraw()`, and `check_balance()` methods.
- Handle negative deposits.
- Handle insufficient funds.
- Handle invalid withdrawal amounts.
- Create a custom `InsufficientFundsError` exception.
- Use the custom exception when a withdrawal exceeds the available balance.

```
class InsufficientFundsError(Exception):
    pass


class BankAccount:
    def __init__(self, balance=0):
        self.balance = balance

    def deposit(self, amount):
        try:
            if amount < 0:
                raise ValueError("Deposit amount cannot be negative.")
            self.balance += amount
            print(f"Deposited: GHS {amount:.2f}")
        except ValueError as e:
            print("Error:", e)

    def withdraw(self, amount):
        try:
            if amount <= 0:
                raise ValueError("Withdrawal amount must be greater than zero.")

            if amount > self.balance:
                raise InsufficientFundsError("Insufficient funds for this withdrawal.")

            self.balance -= amount
            print(f"Withdrawn: GHS {amount:.2f}")

        except ValueError as e:
            print("Error:", e)

        except InsufficientFundsError as e:
            print("Error:", e)

    def check_balance(self):
        print(f"Current Balance: GHS {self.balance:.2f}")


# Create a bank account
account = BankAccount(1000)

# Check initial balance
account.check_balance()

# Make a deposit
account.deposit(500)

# Try a negative deposit
account.deposit(-100)

# Make a valid withdrawal
account.withdraw(300)

# Try to withdraw more than the available balance
account.withdraw(2000)

# Try an invalid withdrawal amount
account.withdraw(0)

# Check final balance
account.check_balance()
```

### Key idea

Exception handling prevents the program from crashing when unexpected or invalid situations occur. Custom exceptions make specific errors easier to understand.
